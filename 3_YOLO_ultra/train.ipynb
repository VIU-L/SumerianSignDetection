```python
from ultralytics import YOLO
```

```python
hyp = {
    'flipud': 0.0,      # vertical flip probability
    'fliplr': 0.0,      # horizontal flip probability
    'degrees': 7.0,     # max rotation degrees
    'translate': 0.1,   # translation factor
    'scale': 0.5,       # scaling factor
    'shear': 7.0,       # shear angle
    'hsv_h': 0.0,     # hue augmentation
    'hsv_s': 0.1,       # saturation augmentation
    'hsv_v': 0.1,       # value augmentation
    'mosaic': 0.8,      # mosaic probability (0–1)
    'mixup': 0.2,       # mixup probability (0–1)
}
```

```python
model = YOLO("YOLO12m.pt")
model.train(
    data="dataset_unicode/dataset.yaml",  # YAML file specifying paths and class names
    epochs=70,            # keep small for a toy run
    imgsz=640,            # resize images to 640x640
    batch=4,              # adjust based on GPU memory
    project="sumerian_yolo",  # folder to save weights and logs
    name="unicode_topN_finalfinal",       # subfolder name
    **hyp,               # data augmentation hyperparameters
    resume = True
)


# 3. Evaluate on validation set (optional)
model.val(data = "../3_YOLO_ultra/dataset_unicode/dataset.yaml", save_json=True, iou = 0.5, conf = 0.1,agnostic_nms=True)

```
